```python
import pandas as pd
import re

df = pd.read_csv(r"/p/project1/westai0091/venturebias/data/output/pitches/phi-4.csv")

def _extract_after_marker(text, marker="explanations.assistant"):
    try:
        if pd.isna(text):
            return text
        if marker in text:
            return text.split(marker)[-1].strip()
        return text
    except Exception:
        return text

def _extract_after_marker2(text, marker="pitch.assistant"):
    try:
        if pd.isna(text):
            return text
        if marker in text:
            return text.split(marker)[-1].strip()
        return text
    except Exception:
        return text

df['clean_response'] = df['clean_response'].apply(_extract_after_marker)
df['clean_response'] = df['clean_response'].apply(_extract_after_marker2)
df['clean_response']

# Transfer "clean_response" to "response" and remove "clean_response"
if 'clean_response' in df.columns:
    df['response'] = df['clean_response']
    df.drop(columns=['clean_response'], inplace=True)


df.to_csv(r"/p/project1/westai0091/venturebias/data/output/pitches/phi-4_copy.csv")
```
